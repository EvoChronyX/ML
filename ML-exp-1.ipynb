```python
def perform_image_eda(file_path):
    """Placeholder function for your extensive Image EDA."""
    print(f"\n[ROUTE] Successfully sent to Image EDA.")
    print(f"-> Processing image file: '{os.path.basename(file_path)}'")



    print("\n[ROUTE] Generalized Image Dataset EDA\n")

    image_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.tiff', '.webp')

    image_paths = []
    labels = []

    dimensions = []
    channels = []
    formats = []
    corrupted = []

    
    # 1. RECURSIVE SCAN (KEY PART)
    
    for root, dirs, files in os.walk(file_path):
        for file in files:
            if file.lower().endswith(image_extensions):

                full_path = os.path.join(root, file)
                label = os.path.basename(os.path.dirname(full_path))

                image_paths.append(full_path)
                labels.append(label)

    print(f"Total Images: {len(image_paths)}")
    print(f"Total Classes: {len(set(labels))}")

    
    # 2. IMAGE VALIDATION
    
    for path in image_paths:
        try:
            with Image.open(path) as img:
                formats.append(img.format)
                dimensions.append(img.size)

                if img.mode == 'RGB':
                    channels.append(3)
                elif img.mode == 'L':
                    channels.append(1)
                else:
                    channels.append(len(img.getbands()))

        except:
            corrupted.append(path)

    
    # 3. STRUCTURAL SUMMARY
    
    print("\nFormat Distribution:")
    print(Counter(formats))
    print("\nUnique Dimensions:")
    print(set(dimensions))
    print("\nChannel Distribution:")
    print(Counter(channels))
    print(f"\n Corrupted Images: {len(corrupted)}")

    
    # 4. LABEL DISTRIBUTION
    
    label_counts = Counter(labels)

    plt.figure(figsize=(12,5))
    plt.bar(label_counts.keys(), label_counts.values())
    plt.title("Class Distribution")
    plt.xticks(rotation=45)
    plt.show()

```

```python
def auto_detect_and_route_dataset(file_path):
    """
    Detects file type based on extension and calls the corresponding EDA function.
    """
    # 1. Check if the file actually exists
    if not os.path.exists(file_path):
        print(f"Error: The file at '{file_path}' does not exist.")
        return

    # 2. Extract and normalize the file extension
    file_extension = Path(file_path).suffix.lower()
    
    # 3. Define accepted extensions
    tabular_extensions = {'.csv', '.tsv', '.xlsx', '.xls', '.json'}

    print(f"Analyzing file: {os.path.basename(file_path)} (Detected extension: {file_extension})")

    # 4. Routing Logic
    if file_extension in tabular_extensions:
        print("System Auto-Detection: Tabular/Structured Data File.")
        perform_tabular_eda(file_path)
        
    else:
        print(" System Auto-Detection: Image File.")
        perform_image_eda(file_path)
        

# ==========================================
# 3. Example Usage
# ==========================================
if __name__ == "__main__":

    auto_detect_and_route_dataset("/home/rohith/SSN/ML/emails.csv")
    print("\n" + "="*40 + "\n")
```
